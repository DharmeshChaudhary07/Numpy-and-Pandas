```python
import numpy as np
```

### Reorganizing Arrays

```python
before = np.array([[1,2,3,4], [5,6,7,8]])
print(before)
print(before.shape)

# Can take a shape unless the no of element in after is same as no of element in before
after = before.reshape((4, 2))
print(after)

after = before.reshape((1,8))
print(after)
```

##### Vertically stackiing vector

```python
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])

print(np.vstack([v2,v1]))
print("---------------")
print(np.vstack([v2,v1,v2,v1]))

# Note that error
# ValueError: all the input array dimensions except for the concatenation axis must match exactly, but along dimension 0,
#  the array at index 0 has size 2 and the array at index 1 has size 3
```

##### Horizontal stacks

```python
h1 = np.ones((2,3))
print(h1)

h2 = np.zeros((2,4))
print(h2)
print("---------------")
print(np.hstack((h1,h2)))

print("---------------")
print(np.hstack((h1,h2,h1,h2)))

# Note that error
# ValueError: all the input array dimensions except for the concatenation axis must match exactly, but along dimension 0,
#  the array at index 0 has size 2 and the array at index 1 has size 3
```

##### Depth Stack

```python
a = np.array([1,3,6,8])
b = np.array([3,6,8,7])

# Stack them along the third dimension
result = np.dstack((a,b))
print(result)
print("New Shape:", result.shape)

print("----------------------")

a = np.array([1,3,6,8])
b = np.array([3,6,8,7])
c = np.array([4,7,8,2])
result = np.dstack((a,b,c))
print(result)
print("New Shape:", result.shape)
```

##### Transpose

```python
transpose = np.array([[1,2,3,5], [3,5,7,2,], [4,6,3,7]])
print(transpose)

print("-------------------")

print(transpose.transpose())
```

##### Split (Vertically & Horizontally)

```python
# Vertical split 
stats = np.array([[1, 2, 3, 4], 
                  [5, 6, 7, 8]])

print(np.vsplit(stats, 2))


# Horizontal split 
stats = np.array([[1, 2, 3, 4], 
                  [5, 6, 7, 8]])

print(np.hsplit(stats, 2))
```

##### Concatenate Array

```python
a = np.array([1,3,6,8])
b = np.array([3,6,8,3])
print(np.concatenate((a,b), axis = 0))
```
