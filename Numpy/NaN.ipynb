### NaN: NumPy vs Pandas

NaN stands for "Not a Number." Think of it simply as a placeholder for missing, blank, or corrupted data in your dataset. It is the programming equivalent of an empty box on a paper form or writing "N/A" (Not Applicable).

```python
import numpy as np
```

```python
a = np.array([1, 2, np.nan])
print(np.sum(a)) 

# The Fix: Use the "Nan-safe" versions
print(np.nansum(a))
```


##### Feature	                                 NumPy (np.nan)	                                        Pandas (NaN / None / PD.NA)

#Primary Rol         A floating-point bit pattern defined by IEEE math standards.	   A data tracking concept representing missing table cells.

#Math Behavior	    Destructive. 1 + np.nan = nan. Breaks normal functions.	Friendly.  Automatically skips blanks during summaries.

#Array Impact	    Forces integers to convert to floats.	                           Tracks them natively using advanced extension types.


NumPy

1. np.nan is a float, so an array containing NaN must be a float array. Integer arrays can’t hold it.

2. NaN spreads through math: np.sum([1, np.nan, 3]) gives nan.

3. Use the NaN-safe versions: np.nansum, np.nanmean, np.nanmax, and so on.

4. NaN never equals itself: np.nan == np.nan is False. Use np.isnan(arr) to check.


Pandas

1. Pandas skips NaN by default: df["col"].sum() and .mean() ignore NaN (skipna=True). This is the biggest everyday difference from NumPy.

2. It has built-in tools: isna(), notna(), fillna(), dropna().

3. Gotcha: an integer column with a missing value becomes float64, so 1, 2, NaN becomes 1.0, 2.0, NaN. To keep integers, use the nullable type Int64 (capital I), which uses pd.NA instead of NaN. This matters when you write IDs to a database or Parquet.

5. In text columns, missing values can show up as None or NaN. isna() treats both as missing.

Rule of thumb: use isna() / fillna() / dropna() on DataFrames, and np.isnan / nan* functions on raw arrays.
