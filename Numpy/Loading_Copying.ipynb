```python
import numpy as np
```

### Loading a File 

```python
filedata = np.genfromtxt('numpydata.txt', delimiter = ',')
print(filedata)
print(filedata.dtype)

print("-----------------------")

filedata = filedata.astype('int32')
print(filedata)
print(filedata.dtype)

print(filedata.shape)
```

##### Boolean Masking And Advanced Indexing

```python
filedata > 50
```

```python
filedata[filedata > 50]
```

```python
## We can index with a list in NumPy
a = np.array([1,4,6,7,3,6,9,4,2,8,6])
print(a.shape)

a[[2,4,8]] # index value returns the element present at index 
```

```python
print(np.any(filedata > 50, axis = 0))
print("------------")
print(np.any(filedata > 50, axis = 1))
```

```python
print(np.all(filedata > 50, axis = 0))
print("------------")
print(np.all(filedata > 50, axis = 1))
```

```python
(~(filedata > 50) & (filedata < 100))
```
