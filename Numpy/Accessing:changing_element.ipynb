Aceessing / Changing specific element, rows, columns etc.

```python
# Initialize Numpy 
import numpy as np
```

Accessing elements

```python
a = np.array([[1,2,3,4,5,6,7,8], [7,6,5,4,3,2,1,0], [1,3,5,6,8,4,9,5]])
print(a)
```

```python
# Get a speciific element (row, column)

print(a[1,5])
print(a[0,7])

print(a[0,-1])          # can you -ve index like list
```

```python
# Get the specific row

print(a[0, :])
print(a[:, 3])
print(a)
```

```python
# Get a [startindex:endindex:steps]
# a[row, startindex:endindex: steps]

# Rows
print(a[0, 1:7 : 2])
print(a[0, 1:-1 : 2])

print(a[0, 0:7 : 2])

print("-------------------")

# Columns 
print(a[:, 0:1 : 1])
print(a[:, 3:4 : 1])
```

Changing Elements

```python
print(a)
print("----------------------------")

a[2,2] = 0
print(a)

print("----------------------------")

a[1,3] = 0
print(a)
```

```python
# Changing the whole Rows and Column

print(a)
print("----------------------------")


a[1, :] = 4
print(a)

print("----------------------------")

a[:, 2] = 0
print(a)
```

3 - D example

```python
b = np.array([ [[1,2], [3,4]] , [[5,6], [7,8]] ])
print(b)
```

```python
# Get Specific element

print(b[0])
print("----------------------------")
print(b[0,1])
print("----------------------------")
print(b[0,1,0])
print(b[0,0,1])

print("----------------------------")

print(b[:,0,:])
print(b[:,1,:])

print("----------------------------")

print(b[0,0,:])
```

Intitalizing Different Types of array

```python
# All 0s martix

print(np.zeros((2,3)))

print("------------------------")

print(np.zeros((4,2,3)))    # 4 dim, 2 rows, 3 columns
```

```python
# All 1s matrix
print(np.ones((4,2,2)))

print("--------------------")

print(np.ones((3,3,4)))             # 3 dim, 3 rows, 4 columns
```

```python
# Any other numbers
np.full((2,3), 12)
# print((np.full((2,3), 12)))
```

```python
# Any other number (full_like)

np.full_like(a, 7)

# take the shape of a variable from above and fill it with 7's 

# or i can use
np.full(a.shape, 6)
```

```python
# Random decimal numbers 

print(np.random.rand(4,3))     # random decimal no of shape 4 * 3

print("--------------------------")

print(np.random.rand(4,3,2))   # 4 dim, 3 rows * 2 col

print("--------------------------")

print(np.random.random_sample(a.shape))

```

```python
# Random Interger Values

print(np.random.randint(7, size=(3,4)))         # 7 is upper bound, random integers between (0,6)

print(np.random.randint(4,7, size=(3,4)))         # between 4,6
```

```python
# Identity Matrix -> An identity matrix is a special square matrix that acts exactly like the number 1 does in regular multiplication.

np.identity(4)
```

```python
# Repeat an array

arr = np.array([1,2,3,4])

repeat1 = np.repeat(arr, 2)
print(repeat1)

repeat2 = np.repeat(arr,3)
print(repeat2)

arr = np.array([[1,2,3,4]])

repeat3 = np.repeat(arr, 4, axis=0)
print(repeat3)
```

Be careful when copying arrays

```python
a1 = np.array([1,2,3,4])
b1 = a1
b1[0] = 7
print(b1)
print(a1)               # also changes in a1

print("---------------------------")

# to prevent that we can use copy function 
a1 = np.array([1,2,3,4])
b1 = a1.copy()
b1[0] = 7
print(b1)
print(a1)               # also changes in a1
```
