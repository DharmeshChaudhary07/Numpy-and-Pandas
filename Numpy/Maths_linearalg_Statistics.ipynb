### Linear Algebra


```python
a = np.ones((2,3))
print(a)

b = np.full((3,2), 2)
print(b)

np.matmul(a,b)              # no of elements in matrix a (row 0) and no of element in  matrix b (column 0 should be same)

# results the output of 2 * 2
```

### Statistics

```python
stats = np.array([[1,2,3],[4,5,6]])
stats
```

```python
print(np.min(stats))

print(np.min(stats, axis = 0))      # np.min(stats, axis=0) (Collapsing Rows / Vertical Action)

print(np.min(stats, axis = 1))      # np.min(stats, axis=1) (Collapsing Columns / Horizontal Action)

```

```python
print(np.max(stats))

print(np.max(stats, axis = 0))      # np.min(stats, axis=0) (Collapsing Rows / Vertical Action)

print(np.max(stats, axis = 1))      # np.min(stats, axis=1) (Collapsing Columns / Horizontal Action)
```

```python
print(np.sum(stats))

print(np.sum(stats, axis = 0))      # np.min(stats, axis=0) (Collapsing Rows / Vertical Action)

print(np.sum(stats, axis = 1))      # np.min(stats, axis=1) (Collapsing Columns / Horizontal Action)
```
