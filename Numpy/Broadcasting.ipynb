### Broadcasting

##### Broadcasting is how NumPy does math between arrays of different shapes without loops.

The rule: compare the shapes from the right-hand side. Two dimensions are compatible if they are equal, or if one of them is 1. A missing dimension counts as 1, and the result takes the larger size.

#### Link Doc: https://numpy.org/doc/stable/user/basics.broadcasting.html

```python
m = np.array([[1, 2, 3],[4, 5, 6]])        # shape (2, 3)
row = np.array([10, 20, 30])
print(row.shape)                           # shape (3,)

print(m + row)        # (2,3) + (3,) -> works, row is applied to every row

print("------------------------")

m = np.array([[1, 2, 3],[4, 5, 6]])        # shape (2, 3)
col = np.array([[100], [200]])   # shape (2, 1)

print(m + col)        # (2,3) + (2,1) -> works, col is applied to every column
```
